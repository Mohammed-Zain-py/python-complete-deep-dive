# What is __name__?

Every Python file has a special variable called __name__.

It tells how the file is being used:

* "__main__" → when file is run directly
* module name → when imported

# Basic Example

```python
print(__name__)
```

# Using __name__ == "__main__"

```python
def main():
    print("Running main function")

if __name__ == "__main__":
    main()
```

# Why is This Important?

It ensures:

* Code runs only when file is executed directly
* Prevents execution when imported

# Example with Import

```python
# create a file named file1.py

# andd put the below content in it

def greet():
    print("Hello")

if __name__ == "__main__":
    greet()
```

```python
# file2.py

import file1

print("Imported file1")
```
