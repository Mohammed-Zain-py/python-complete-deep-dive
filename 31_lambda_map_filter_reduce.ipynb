# Lambda Functions

A lambda function is a **small anonymous function**.

Syntax:

lambda arguments: expression

```python
# Normal function
def square(x):
    return x * x

# Lambda function
square_lambda = lambda x: x * x

print(square_lambda(5))
```

# map() Function

map() applies a function to all elements of an iterable.

```python
nums = [1, 2, 3, 4]

result = list(map(lambda x: x * x, nums))

print(result)
```

# filter() Function

filter() selects elements based on a condition.

```python
nums = [1, 2, 3, 4, 5, 6]

result = list(filter(lambda x: x % 2 == 0, nums))

print(result)
```

# reduce() Function

reduce() applies a function cumulatively.

It is available in functools module.

```python
from functools import reduce

nums = [1, 2, 3, 4]

result = reduce(lambda x, y: x + y, nums)

print(result)
```

# Practice Problems

```python
# Problem 1: Square all elements using map

nums = [1, 2, 3, 4]

print(list(map(lambda x: x * x, nums)))
```

```python
# Problem 2: Filter odd numbers

nums = [1, 2, 3, 4, 5]

print(list(filter(lambda x: x % 2 != 0, nums)))
```

```python
# Problem 3: Find product using reduce

from functools import reduce

nums = [1, 2, 3, 4]

print(reduce(lambda x, y: x * y, nums))
```

```python
# Problem 4: Convert list of strings to uppercase

words = ["python", "code"]

print(list(map(lambda x: x.upper(), words)))
```
