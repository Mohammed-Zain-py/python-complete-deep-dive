## Day 52 — Mini Project: Number Pattern Analyzer

In this notebook, we will:

* analyze a list of numbers
* count even and odd numbers
* find duplicates
* identify increasing / decreasing trend
* generate summary insights

# Why This Project?

This project improves:

* loop logic
* condition handling
* dictionary counting
* DSA thinking

```python
numbers = [2, 4, 6, 3, 5, 6, 8, 2, 10]
```

# Step 1: Count Even and Odd Numbers

```python
even_count = 0
odd_count = 0

for num in numbers:
    if num % 2 == 0:
        even_count += 1
    else:
        odd_count += 1

print("Even:", even_count)
print("Odd:", odd_count)
```

# Step 2: Find Duplicates

```python
frequency = {}

for num in numbers:
    frequency[num] = frequency.get(num, 0) + 1

duplicates = [num for num, count in frequency.items() if count > 1]

print("Duplicates:", duplicates)
```

# Step 3: Check Trend

```python
is_increasing = all(numbers[i] <= numbers[i + 1] for i in range(len(numbers) - 1))

print("Increasing Order:", is_increasing)
```

# Step 4: Summary Statistics

```python
print("Min:", min(numbers))
print("Max:", max(numbers))
print("Sum:", sum(numbers))
print("Average:", sum(numbers) / len(numbers))
```
