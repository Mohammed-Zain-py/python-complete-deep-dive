# What is Inheritance?

Inheritance allows a class to **use properties and methods of another class**.

Benefits:

* Code reuse
* Cleaner code
* Better structure

# Single Inheritance

One child class inherits from one parent class.

```python
class Animal:
    def speak(self):
        print("Animal speaks")

class Dog(Animal):
    pass

d = Dog()
d.speak()
```

# Adding Child Methods

```python
class Animal:
    def speak(self):
        print("Animal speaks")

class Dog(Animal):
    def bark(self):
        print("Dog barks")

d = Dog()
d.speak()
d.bark()
```

# Multiple Inheritance

A class inherits from more than one parent class.

```python
class Father:
    def skills(self):
        print("Coding")

class Mother:
    def talent(self):
        print("Drawing")

class Child(Father, Mother):
    pass

c = Child()
c.skills()
c.talent()
```

# Multilevel Inheritance

A class inherits from another class which also inherits from another class.

```python
class Grandparent:
    def house(self):
        print("Grandparent's house")

class Parent(Grandparent):
    pass

class Child(Parent):
    pass

c = Child()
c.house()
```

# Practice Problems

```python
# Problem 1: Create parent and child class

class Vehicle:
    def start(self):
        print("Vehicle started")

class Car(Vehicle):
    pass

c = Car()
c.start()
```

```python
# Problem 2: Add method in child class

class Animal:
    def eat(self):
        print("Eating")

class Dog(Animal):
    def bark(self):
        print("Barking")

d = Dog()
d.eat()
d.bark()
```

```python
# Problem 3: Multiple inheritance example

class A:
    def method1(self):
        print("A")

class B:
    def method2(self):
        print("B")

class C(A, B):
    pass

obj = C()
obj.method1()
obj.method2()
```

```python
# Problem 4: Multilevel inheritance example

class A:
    def method(self):
        print("A")

class B(A):
    pass

class C(B):
    pass

obj = C()
obj.method()
```
