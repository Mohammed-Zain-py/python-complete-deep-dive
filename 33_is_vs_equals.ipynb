## Day 33 — is vs == in Python

In this notebook, we will cover:

* Difference between is and ==
* Identity vs equality
* Memory reference concept
* Practice examples

# == Operator (Equality)

== checks whether **values are equal**.

```python
a = 10
b = 10

print(a == b)  # True
```

# is Operator (Identity)

is checks whether **both variables point to the same object in memory**.

```python
a = 10
b = 10

print(a is b)  # True (small integers are cached)
```

# Important Example (Lists)

```python
a = [1, 2, 3]
b = [1, 2, 3]

print(a == b)  # True (same values)
print(a is b)  # False (different objects)
```

# Same Reference Example

```python
a = [1, 2, 3]
b = a

print(a == b)  # True
print(a is b)  # True (same object)
```

# Key Difference

* == → compares values
* is → compares memory location

# Practice Problems

```python
# Problem 1

a = [10, 20]
b = [10, 20]

print(a == b)
print(a is b)
```

```python
# Problem 2

x = 1000
y = 1000

print(x == y)
print(x is y)
```

```python
# Problem 3

a = None
b = None

print(a is b)   # recommended usage
```

# Summary

In this notebook, we learned:

* == checks value equality
* is checks object identity
* Memory reference matters
* Common interview traps

Always use `is` for comparing with None.
