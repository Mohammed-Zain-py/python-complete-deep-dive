# Importing re Module

```python
import re
```

# search() Function

Finds first match.

```python
text = "Python is powerful"

result = re.search("power", text)

print(result)
```

# findall() Function

Finds all matches.

```python
text = "cat bat mat"

print(re.findall("at", text))
```

# match() Function

Matches from beginning of string.

```python
text = "Python"

print(re.match("Py", text))
print(re.match("th", text))
```

# Meta Characters

* . → any character
* ^ → start of string
* $ → end of string
* * → 0 or more
* + → 1 or more
* ? → 0 or 1
* [] → set of characters

```python
text = "cat bat rat"

print(re.findall("[cbr]at", text))
```

# Practice Problems

```python
# Problem 1: Find all digits

text = "My number is 12345"

print(re.findall("\d", text))
```

```python
# Problem 2: Find all words starting with 'p'

text = "python is powerful and popular"

print(re.findall("p\w+", text))
```

```python
# Problem 3: Validate email pattern

text = "test@gmail.com"

print(re.match("\w+@\w+\.\w+", text))
```

```python
# Problem 4: Extract numbers

text = "Order 123, price 456"

print(re.findall("\d+", text))
```
