## Day 48 — Project Upgrade: Log Analyzer with Regex

In this notebook, we will:

* Improve the previous log analyzer
* Use regex for better parsing
* Extract timestamps and log levels
* Generate useful insights

# Why Upgrade This?

Instead of simple string checks, we now use **regex-based parsing**.

This makes the project:

* More scalable
* More realistic
* More impressive

```python
import re
```

# Step 1: Sample Logs

```python
logs = [
    "2026-02-22 10:00:01 ERROR Database connection failed",
    "2026-02-22 10:02:15 WARNING Low memory",
    "2026-02-22 10:03:40 INFO Application started",
    "2026-02-22 10:05:22 ERROR Timeout occurred"
]
```

# Step 2: Regex Pattern

```python
pattern = r"(\d{4}-\d{2}-\d{2})\s(\d{2}:\d{2}:\d{2})\s(\w+)\s(.+)"
```

# Step 3: Parse Logs

```python
parsed_logs = []

for log in logs:
    match = re.match(pattern, log)
    
    if match:
        date, time, level, message = match.groups()
        
        parsed_logs.append({
            "date": date,
            "time": time,
            "level": level,
            "message": message
        })

print(parsed_logs)
```

# Step 4: Count Log Levels

```python
counts = {}

for log in parsed_logs:
    level = log["level"]
    counts[level] = counts.get(level, 0) + 1

print(counts)
```

# Step 5: Extract Only Errors

```python
errors = [log for log in parsed_logs if log["level"] == "ERROR"]

print(errors)
```
