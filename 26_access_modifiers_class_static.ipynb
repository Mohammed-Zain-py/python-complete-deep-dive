# Access Modifiers in Python

Access modifiers control **visibility of variables and methods**.

Types:

* Public
* Protected
* Private

# Public Members

Accessible from anywhere.# Public Members

Accessible from anywhere.

```python
class Student:
    def __init__(self):
        self.name = "Zain"   # public

s = Student()
print(s.name)
```

# Protected Members

Use single underscore (_).

Used as a convention (not strict).

```python
class Student:
    def __init__(self):
        self._age = 21

s = Student()
print(s._age)
```

# Private Members

Use double underscore (__).

Name mangling is applied.

```python
class Student:
    def __init__(self):
        self.__marks = 95

s = Student()

# print(s.__marks)  # will give error
print(s._Student__marks)  # accessing private
```

# Class Variables vs Instance Variables

* Instance variable → unique for each object
* Class variable → shared among all objects

```python
class Student:
    school = "ABC School"   # class variable

    def __init__(self, name):
        self.name = name   # instance variable

s1 = Student("Zain")
s2 = Student("Ali")

print(s1.school)
print(s2.school)
```

# Class Methods

Used to access class variables.

Defined using @classmethod.

```python
class Student:
    school = "ABC School"

    @classmethod
    def get_school(cls):
        return cls.school

print(Student.get_school())
```

# Static Methods

Used when method does not depend on class or object.

```python
class Math:
    @staticmethod
    def add(a, b):
        return a + b

print(Math.add(5, 3))
```

```python
# Problem 1: Class variable example

class Company:
    name = "Google"

c1 = Company()
c2 = Company()

print(c1.name)
print(c2.name)
```

```python
# Problem 2: Class method example

class Test:
    value = 10

    @classmethod
    def show(cls):
        print(cls.value)

Test.show()
```

```python
# Problem 3: Static method example

class Calculator:
    @staticmethod
    def multiply(a, b):
        return a * b

print(Calculator.multiply(4, 5))
```
