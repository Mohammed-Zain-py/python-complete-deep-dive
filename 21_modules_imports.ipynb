# What are Modules?

A module is a **file containing Python code** (functions, variables, classes).

Modules help in:

* Code reuse
* Better organization
* Avoiding repetition

# Importing a Module

We use the `import` keyword.

```python
import math

print(math.sqrt(16))
```

# Import Specific Functions

```python
from math import sqrt

print(sqrt(25))
```

# Import with Alias

```python
import math as m

print(m.sqrt(36))
```

# Import Everything (Not Recommended)

```python
from math import *

print(sqrt(49))
```

# Built-in Modules

```python
import random

print(random.randint(1, 10))
```

# Practice Problems

```python
# Problem 1: Find square root using math module

import math

print(math.sqrt(64))
```

```python
# Problem 2: Generate random number

import random

print(random.randint(1, 100))
```
