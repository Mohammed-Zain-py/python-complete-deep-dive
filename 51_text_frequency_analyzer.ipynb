## Day 51 — Mini Project: Text Frequency Analyzer

In this notebook, we will:

* Analyze word frequency
* Count repeated words
* Find most frequent word
* Generate summary

# Why This Project?

This project improves:

* dictionary usage
* hashing logic
* data analysis thinking

Very useful before DSA starts.

```python
text = "python is easy and python is powerful and python is fast"
```

# Step 1: Split Words

```python
words = text.split()

print(words)
```

# Step 2: Count Frequency

```python
frequency = {}

for word in words:
    frequency[word] = frequency.get(word, 0) + 1

print(frequency)
```

# Step 3: Most Frequent Word

```python
most_frequent = max(frequency, key=frequency.get)

print("Most Frequent Word:", most_frequent)
print("Count:", frequency[most_frequent])
```

# Step 4: Sorted Frequency

```python
sorted_freq = sorted(
    frequency.items(),
    key=lambda x: x[1],
    reverse=True
)

print(sorted_freq)
```

# Summary

In this notebook, we:

* Counted word frequency
* Found repeated words
* Identified most frequent word
* Sorted results
