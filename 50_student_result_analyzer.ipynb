## Day 50 — Mini Project: Student Result Analyzer

In this notebook, we will:

* Analyze student marks
* Calculate average score
* Find topper
* Generate pass/fail summary

# Why This Project?

This project demonstrates:

* Data analysis
* Aggregation
* Reporting
* Summary generation

```python
students = [
    {"name": "Zain", "marks": 85},
    {"name": "Ali", "marks": 72},
    {"name": "John", "marks": 91},
    {"name": "Sara", "marks": 40}
]
```

# Step 1: Calculate Average Marks

```python
average = sum(student["marks"] for student in students) / len(students)

print("Average Marks:", average)
```

# Step 2: Find Topper

```python
topper = max(students, key=lambda x: x["marks"])

print("Topper:", topper["name"], topper["marks"])
```

# Step 3: Pass / Fail Summary

```python
passed = []
failed = []

for student in students:
    if student["marks"] >= 40:
        passed.append(student["name"])
    else:
        failed.append(student["name"])

print("Passed:", passed)
print("Failed:", failed)
```

# Step 4: Grade Assignment

```python
for student in students:
    marks = student["marks"]
    
    if marks >= 90:
        grade = "A"
    elif marks >= 75:
        grade = "B"
    elif marks >= 50:
        grade = "C"
    else:
        grade = "D"
    
    print(student["name"], grade)
```

* Calculated average
* Found topper

* Assigned grades
