## Day 46 — Project: API to CSV Converter

In this notebook, we will:

* Fetch data from an API
* Process JSON data
* Convert it into CSV format
* Save it for analysis

# Why This Project?

This connects to:

* Data Analytics

* CSV datasets for analysis

```python
import requests
import csv
```

# Step 1: Fetch API Data

```python
url = "https://jsonplaceholder.typicode.com/users"

response = requests.get(url)
data = response.json()

print(data[:2])   # preview
```

# Step 2: Extract Required Fields

```python
users = []

for item in data:
    users.append({
        "id": item["id"],
        "name": item["name"],
        "email": item["email"],
        "city": item["address"]["city"]
    })

print(users[:2])
```

# Step 3: Write to CSV

```python
with open("users.csv", "w", newline="") as file:
    fieldnames = ["id", "name", "email", "city"]

    writer = csv.DictWriter(file, fieldnames=fieldnames)
    
    writer.writeheader()
    writer.writerows(users)
```

# Step 4: Read CSV (Verification)

```python
with open("users.csv", "r") as file:
    reader = csv.reader(file)
    
    for row in reader:
        print(row)
```
