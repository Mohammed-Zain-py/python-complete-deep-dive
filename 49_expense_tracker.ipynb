## Day 49 — Mini Project: Expense Tracker

In this notebook, we will:

* Record expenses
* Categorize them
* Calculate total spend
* Generate a simple summary

# Why This Project?

Expense tracking is a real-world use case.

It shows:

* data storage
* aggregation
* reporting

```python
expenses = [
    {"category": "Food", "amount": 200},
    {"category": "Travel", "amount": 150},
    {"category": "Food", "amount": 100},
    {"category": "Shopping", "amount": 500}
]
```

# Step 1: Calculate Total Expense

```python
total = sum(item["amount"] for item in expenses)

print("Total Expense:", total)
```

# Step 2: Category-wise Summary

```python
summary = {}

for item in expenses:
    category = item["category"]
    amount = item["amount"]
    
    summary[category] = summary.get(category, 0) + amount

print(summary)
```

# Step 3: Highest Expense Category

```python
highest = max(summary, key=summary.get)

print("Highest Spending Category:", highest)
```

# Step 4: Add New Expense

```python
new_expense = {"category": "Bills", "amount": 300}

expenses.append(new_expense)

print(expenses)
```
