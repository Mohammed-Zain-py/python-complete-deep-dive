# What is a CLI Tool?

A CLI (Command Line Interface) tool allows users to **run programs using terminal commands**.

Examples:

* git
* python
* pip

We can build our own CLI tools using Python.

# Using argparse (Recommended)

```python
import argparse

parser = argparse.ArgumentParser()

parser.add_argument("name", help="Enter your name")

args = parser.parse_args(['Zain'])

print("Hello", args.name)
```

# Adding Multiple Arguments

```python
import argparse

parser = argparse.ArgumentParser()

parser.add_argument("a", type=int)
parser.add_argument("b", type=int)

args = parser.parse_args(['5','3'])

print("Sum:", args.a + args.b)
```

# Optional Arguments

```python
import argparse

parser = argparse.ArgumentParser()

parser.add_argument("--name", help="Your name")

args = parser.parse_args(['--name', 'Zain'])

print(args.name)
```

# Practice: Simple Calculator CLI

```python
import argparse

parser = argparse.ArgumentParser()

parser.add_argument("a", type=int)
parser.add_argument("b", type=int)
parser.add_argument("operation", help="add/sub")

# args = parser.parse_args(['10', '5', 'add'])
args = parser.parse_args(['10', '5', 'sub'])

if args.operation == "add":
    print(args.a + args.b)
elif args.operation == "sub":
    print(args.a - args.b)
```
