# What is OOP?

Object Oriented Programming (OOP) is a way of writing code using **objects and classes**.

It helps in:

* Organizing code
* Reusability
* Real-world modeling

# What is a Class?

A class is a **blueprint for creating objects**.

```python
class Student:
    pass
```

# What is an Object?

An object is an **instance of a class**.

```python
class Student:
    pass

s1 = Student()
s2 = Student()

print(s1)
print(s2)
```

# __init__ Constructor

The constructor is used to **initialize object properties**.

```python
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

s1 = Student("Zain", 21)

print(s1.name)
print(s1.age)
```

# Adding Methods

```python
class Student:
    def __init__(self, name):
        self.name = name

    def greet(self):
        print("Hello", self.name)

s1 = Student("Zain")
s1.greet()
```

# Practice Problems

```python
# Problem 1: Create a class for Car

class Car:
    def __init__(self, brand):
        self.brand = brand

car1 = Car("Toyota")
print(car1.brand)
```

```python
# Problem 2: Add method to print info

class Person:
    def __init__(self, name):
        self.name = name

    def show(self):
        print("Name:", self.name)

p1 = Person("Zain")
p1.show()
```

```python
# Problem 3: Create class for Rectangle

class Rectangle:
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width

r = Rectangle(4, 5)
print(r.area())
```
