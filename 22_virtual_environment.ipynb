# What is a Virtual Environment?

A virtual environment is an **isolated environment for Python projects**.

It allows each project to have its own:

* Python version
* Libraries
* Dependencies

# Why Use Virtual Environment?

Without virtual environment:

* All projects share same libraries
* Version conflicts may occur

With virtual environment:

* Each project is isolated
* No conflicts
* Clean project setup

# Creating a Virtual Environment

Use the following command in the 'terminal' :

```python
python -m venv myenv
```

# Activating Virtual Environment

Windows:

```python
myenv\Scripts\activate
```

# Installing Packages

```python
pip install numpy
```

# requirements.txt

Used to store project dependencies.

Command to create:


```python
pip freeze > requirements.txt
```

# Installing from requirements.txt

```python
pip install -r requirements.txt
```

# Key Points

* Virtual environments isolate projects
* Avoid dependency conflicts
* requirements.txt helps share dependencies
