# What are Generators?

Generators are functions that **return values one at a time using yield** instead of returning all at once.

They are used for:

* Memory efficiency
* Working with large data

# What are Generators?

Generators are functions that **return values one at a time using yield** instead of returning all at once.

They are used for:

* Memory efficiency
* Working with large data

```python
def my_generator():
    yield 1
    yield 2
    yield 3

g = my_generator()

print(next(g))
print(next(g))
print(next(g))
```

# Generator with Loop

```python
def count_up(n):
    for i in range(1, n + 1):
        yield i

for num in count_up(5):
    print(num)
```

# Generator vs Normal Function

```python
# Normal function

def normal():
    return [1, 2, 3]

# Generator

def gen():
    yield 1
    yield 2
    yield 3

print(normal())
print(list(gen()))
```

# Benefits of Generators

* Saves memory
* Faster for large data
* Lazy evaluation (produces values when needed)

# Practice Problems

```python
# Problem 1: Generate numbers from 1 to n

def generate(n):
    for i in range(1, n + 1):
        yield i

for num in generate(5):
    print(num)
```

```python
# Problem 2: Generate squares

def squares(n):
    for i in range(1, n + 1):
        yield i * i

for num in squares(5):
    print(num)
```

```python
# Problem 3: Generate even numbers

def even_numbers(n):
    for i in range(n):
        if i % 2 == 0:
            yield i

for num in even_numbers(10):
    print(num)
```
