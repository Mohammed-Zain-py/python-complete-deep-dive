# What is Recursion?

Recursion is a technique where a function **calls itself** to solve a problem.

It is used to break a problem into **smaller subproblems**.

Every recursive function must have:

* Base Case → stops recursion
* Recursive Case → calls itself

# Basic Example

```python
def count_down(n):
    if n == 0:
        return
    print(n)
    count_down(n - 1)

count_down(5)
```

# Factorial using Recursion

```python
def factorial(n):
    if n == 0:
        return 1
    return n * factorial(n - 1)

print(factorial(5))
```

# Fibonacci using Recursion# Fibonacci using Recursion

```python
def fibonacci(n):
    if n <= 1:
        return n
    return fibonacci(n - 1) + fibonacci(n - 2)

print(fibonacci(6))
```

# Key Points

* Always define a base case
* Avoid infinite recursion
* Recursion can be less efficient than loops in some cases

# Practice Problems

```python
# Problem 1: Print numbers from n to 1

def print_numbers(n):
    if n == 0:
        return
    print(n)
    print_numbers(n - 1)

print_numbers(5)
```

```python
# Problem 2: Sum of first n numbers

def sum_n(n):
    if n == 0:
        return 0
    return n + sum_n(n - 1)

print(sum_n(5))
```

```python
# Problem 3: Power of a number

def power(a, b):
    if b == 0:
        return 1
    return a * power(a, b - 1)

print(power(2, 3))
```

```python
# Problem 4: Reverse a string using recursion

def reverse_string(s):
    if s == "":
        return s
    return reverse_string(s[1:]) + s[0]

print(reverse_string("python"))
```
