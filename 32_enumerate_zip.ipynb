# enumerate() Function

enumerate() is used to get both:

* index
* value

while looping.

```python
names = ["Zain", "Ali", "John"]

for index, value in enumerate(names):
    print(index, value)
```

```python
# Starting index from 1

names = ["Zain", "Ali"]

for i, name in enumerate(names, start=1):
    print(i, name)
```

# zip() Function

zip() combines multiple iterables.

```python
names = ["Zain", "Ali"]
marks = [90, 85]

result = list(zip(names, marks))

print(result)
```

```python
for name, mark in zip(names, marks):
    print(name, mark)
```

# Unzipping

```python
pairs = [("Zain", 90), ("Ali", 85)]

names, marks = zip(*pairs)

print(names)
print(marks)
```

# Practice Problems

```python
# Problem 1: Print index and element

nums = [10, 20, 30]

for i, val in enumerate(nums):
    print(i, val)
```

```python
# Problem 2: Combine two lists

a = [1, 2, 3]
b = [4, 5, 6]

print(list(zip(a, b)))
```

```python
# Problem 3: Create dictionary from two lists

keys = ["name", "age"]
values = ["Zain", 21]

d = dict(zip(keys, values))

print(d)
```

```python
# Problem 4: Loop through two lists

names = ["Zain", "Ali"]
marks = [90, 80]

for name, mark in zip(names, marks):
    print(f"{name} scored {mark}")
```
