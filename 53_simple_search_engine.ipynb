## Day 53 — Mini Project: Simple Search Engine

In this notebook, we will:

* search for keywords in text
* count keyword matches
* return matching lines
* build search logic

# Why This Project?

This project improves:

* string searching
* text processing
* loop + condition logic

Excellent bridge to DSA string problems.

```python
documents = [
    "Python is easy to learn",
    "DSA improves problem solving",
    "Python is widely used in data science",
    "String searching is important in DSA"
]
```

# Step 1: Define Search Keyword

```python
keyword = "Python"
```

# Step 2: Find Matching Documents

```python
matches = []

for doc in documents:
    if keyword.lower() in doc.lower():
        matches.append(doc)

print(matches)
```

# Step 3: Count Matches

```python
print("Total Matches:", len(matches))
```

# Step 4: Word Frequency in Matches

```python
frequency = {}

for doc in matches:
    for word in doc.split():
        word = word.lower()
        frequency[word] = frequency.get(word, 0) + 1

print(frequency)
```
