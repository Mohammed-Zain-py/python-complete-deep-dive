# What are Exceptions?

An exception is an **error that occurs during program execution**.

Example errors:

* Division by zero
* Invalid input
* File not found

Exception handling allows us to **handle errors gracefully**.

# try and except

We use try and except to catch errors.

```python
try:
    x = 10 / 0
except:
    print("Error occurred")
```

# Specific Exceptions

```python
try:
    num = int("abc")
except ValueError:
    print("Invalid input")
```

# else Block

The else block runs if no exception occurs.

```python
try:
    x = 10 / 2
except:
    print("Error")
else:
    print("No error occurred")
```

# finally Block

The finally block always executes.

```python
try:
    x = 10 / 2
except:
    print("Error")
finally:
    print("Execution completed")
```

# Raising Custom Errors

```python
age = -5

if age < 0:
    raise ValueError("Age cannot be negative")
```

# Practice Problems

```python
# Problem 1: Handle division error

try:
    a = 10
    b = 0
    print(a / b)
except ZeroDivisionError:
    print("Cannot divide by zero")
```

```python
# Problem 2: Handle invalid input

try:
    num = int("abc")
except ValueError:
    print("Invalid number")
```

```python
# Problem 3: Use finally block

try:
    print("Trying...")
finally:
    print("Always runs")
```

```python
# Problem 4: Raise error if number is negative

num = -10

if num < 0:
    raise ValueError("Negative number not allowed")
```
