# String Concatenation

We can combine strings using the `+` operator.

```python
name = "Zain"
age = 21

print("My name is " + name + " and I am " + str(age) + " years old")
```

# format() Method

The format() method is used to insert values into a string.

```python
name = "Zain"
age = 21

print("My name is {} and I am {} years old".format(name, age))
```

```python
# Using index positions

print("My name is {0} and I am {1} years old".format("Zain", 21))
```

# f-Strings (Recommended)

f-strings are the modern and most readable way of formatting strings.

```python
name = "Zain"
age = 21

print(f"My name is {name} and I am {age} years old")
```

```python
# Using expressions

a = 5
b = 3

print(f"Sum is {a + b}")
```

# Practice Problems

```python
# Problem 1: Print name and city

name = "Zain"
city = "Bangalore"

print(f"My name is {name} and I live in {city}")
```

```python
# Problem 2: Print result of multiplication

a = 4
b = 6

print(f"Multiplication of {a} and {b} is {a * b}")
```

```python
# Problem 3: Print formatted sentence using format()

name = "Zain"
marks = 95

print("Student {} scored {} marks".format(name, marks))
```
