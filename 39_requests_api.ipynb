# Importing requests

```python
import requests
```

# GET Request

```python
response = requests.get("https://jsonplaceholder.typicode.com/posts/1")

print(response.status_code)
print(response.text)
```

# JSON Response

```python
data = response.json()

print(data)
print(data["title"])
```

# POST Request

```python
url = "https://jsonplaceholder.typicode.com/posts"

payload = {
    "title": "Python",
    "body": "Learning API",
    "userId": 1
}

response = requests.post(url, json=payload)

print(response.status_code)
print(response.json())
```
