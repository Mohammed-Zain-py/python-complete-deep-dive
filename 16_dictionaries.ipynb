# What are Dictionaries?

A dictionary is a **collection of key-value pairs**.

It is used to store data in the form:

key → value

Dictionaries are:

* Unordered (Python 3.7+ maintains insertion order)
* Mutable
* Keys must be unique

```python
student = {
    "name": "Zain",
    "age": 21,
    "marks": 95
}

print(student)
```

# Accessing Elements

```python
student = {
    "name": "Zain",
    "age": 21
}

print(student["name"])
print(student["age"])
```

# Using get() Method

```python
student = {"name": "Zain"}

print(student.get("name"))
print(student.get("age"))   # returns None instead of error
```

# Adding and Updating Elements

```python
student = {"name": "Zain"}

student["age"] = 21   # add
student["name"] = "Ali"  # update

print(student)
```

# Removing Elements

```python
student = {"name": "Zain", "age": 21}

student.pop("age")
print(student)
```

# Dictionary Methods

```python
student = {"name": "Zain", "age": 21}

print(student.keys())
print(student.values())
print(student.items())
```

# Looping Through Dictionary

```python
student = {"name": "Zain", "age": 21}

for key in student:
    print(key, student[key])
```

# Practice Problems

```python
# Problem 1: Print all keys

student = {"name": "Zain", "age": 21}

for key in student:
    print(key)
```

```python
# Problem 2: Print all values

student = {"name": "Zain", "age": 21}

for value in student.values():
    print(value)
```

```python
# Problem 3: Count frequency of elements

nums = [1, 2, 2, 3, 3, 3]

freq = {}

for num in nums:
    if num in freq:
        freq[num] += 1
    else:
        freq[num] = 1

print(freq)
```

```python
# Problem 4: Check if key exists

student = {"name": "Zain", "age": 21}

print("name" in student)
print("marks" in student)
```

```python
# Problem 5: Merge two dictionaries

a = {"x": 1}
b = {"y": 2}

a.update(b)

print(a)
```
