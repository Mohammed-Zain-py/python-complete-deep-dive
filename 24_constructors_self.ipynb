# What is a Constructor?

A constructor is a special method used to **initialize objects**.

In Python, we use:

__init__()

# Default Constructor

```python
class Student:
    def __init__(self):
        print("Constructor called")

s1 = Student()
```

# Parameterized Constructor

```python
class Student:
    def __init__(self, name):
        self.name = name

s1 = Student("Zain")
print(s1.name)
```

# self Keyword

self represents the **current object**.

It is used to access:

* instance variables
* methods

```python
class Student:
    def __init__(self, name):
        self.name = name

    def show(self):
        print(self.name)

s1 = Student("Zain")
s1.show()
```

# Why self is Important

Without self:

* Python cannot differentiate between variables
* Object data cannot be stored properly

# Practice Problems

```python
# Problem 1: Create class with default constructor

class Test:
    def __init__(self):
        print("Hello from constructor")

t = Test()
```

```python
# Problem 2: Create class with parameterized constructor

class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

p = Person("Zain", 21)
print(p.name, p.age)
```

```python
# Problem 3: Print student details using method

class Student:
    def __init__(self, name):
        self.name = name

    def display(self):
        print("Name:", self.name)

s = Student("Zain")
s.display()
```

```python
# Problem 4: Create class for Circle

class Circle:
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return 3.14 * self.radius * self.radius

c = Circle(5)
print(c.area())
```
