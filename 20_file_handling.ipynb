# What is File Handling?

File handling is used to **read, write, and manipulate files**.

Python provides built-in functions to work with files.

# Opening a File

We use the `open()` function.

Syntax:

open("filename", "mode")

```python
file = open("sample.txt", "r")
```

# File Modes

* "r" → Read (default)
* "w" → Write (overwrites file)
* "a" → Append
* "x" → Create new file

# Reading from a File

#### Note : create a file named sample.txt before exicuting the below cells

```python
file = open("sample.txt", "r")

content = file.read()
print(content)

file.close()
```

```python
file = open("sample.txt", "r")

print(file.readline())

file.close()
```

# Writing to a File

```python
file = open("sample.txt", "w")

file.write("Hello, this is Python")

file.close()
```

```python
file = open("sample.txt", "a")

file.write("\nAppending new line")

file.close()
```

# Using with Statement (Recommended)

Automatically closes the file.

```python
with open("sample.txt", "r") as file:
    content = file.read()
    print(content)
```

# Practice Problems

```python
# Problem 1: Write text to a file

with open("test.txt", "w") as file:
    file.write("Python is easy")
```

```python
# Problem 2: Read content from file

with open("test.txt", "r") as file:
    print(file.read())
```

```python
# Problem 3: Append text to file

with open("test.txt", "a") as file:
    file.write("\nLearning file handling")
```

```python
# Problem 4: Read line by line

with open("test.txt", "r") as file:
    for line in file:
        print(line.strip())
```
