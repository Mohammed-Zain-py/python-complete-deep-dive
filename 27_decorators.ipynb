# What are Decorators?

Decorators are used to **modify or extend the behavior of functions** without changing their code.

They are written using @ syntax.

# Functions as Objects

In Python, functions can be:

* Passed as arguments
* Returned from other functions

```python
def greet():
    print("Hello")

func = greet
func()
```

# Creating a Simple Decorator

```python
def decorator(func):
    def wrapper():
        print("Before function")
        func()
        print("After function")
    return wrapper
```

# Using the Decorator

```python
def say_hello():
    print("Hello")

decorated = decorator(say_hello)
decorated()
```

# Using @ Syntax (Recommended)

```python
@decorator
def greet():
    print("Hi")

greet()
```

# Decorator with Arguments

```python
def decorator(func):
    def wrapper(a, b):
        print("Before function")
        result = func(a, b)
        print("After function")
        return result
    return wrapper

@decorator
def add(a, b):
    return a + b

print(add(2, 3))
```

# Practice Problems

```python
# Problem 1: Create decorator to print start and end

def my_decorator(func):
    def wrapper():
        print("Start")
        func()
        print("End")
    return wrapper

@my_decorator
def test():
    print("Inside function")

test()
```

```python
# Problem 2: Decorator with arguments

def decorator(func):
    def wrapper(x):
        print("Processing...")
        return func(x)
    return wrapper

@decorator
def square(x):
    return x * x

print(square(4))
```
