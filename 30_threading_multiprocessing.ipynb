# What is Threading?

Threading allows a program to **run multiple tasks at the same time**.

Used for:

* I/O tasks
* Faster execution

# Creating a Thread

```python
import threading

def task():
    print("Thread is running")

t1 = threading.Thread(target=task)
t1.start()
t1.join()
```

# Multiple Threads

```python
import threading

def task(n):
    print("Task", n)

threads = []

for i in range(3):
    t = threading.Thread(target=task, args=(i,))
    threads.append(t)
    t.start()

for t in threads:
    t.join()
```

# What is Multiprocessing?

Multiprocessing uses **multiple CPU cores** to run tasks in parallel.

Better for CPU-heavy tasks.

# Creating a Process

```python
import multiprocessing

def task():
    print("Process running")

p = multiprocessing.Process(target=task)
p.start()
p.join()
```

# Multiple Processes

```python
import multiprocessing

def task(n):
    print("Process", n)

processes = []

for i in range(3):
    p = multiprocessing.Process(target=task, args=(i,))
    processes.append(p)
    p.start()

for p in processes:
    p.join()
```

# Threading vs Multiprocessing

* Threading → same memory space, faster for I/O
* Multiprocessing → separate memory, better for CPU tasks
