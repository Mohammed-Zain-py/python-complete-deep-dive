# What is a Log Analyzer?

A log analyzer processes log files to:

* Detect errors
* Count warnings
* Extract useful insights

Used in:

* Servers
* Applications
* Debugging systems

# Step 1: Sample Log File

Create a file named logs.txt with content like:

INFO: Application started  
ERROR: Failed to connect database  
WARNING: Low memory  
ERROR: Timeout occurred  
INFO: Process completed

# Step 2: Read Log File

```python
with open("logs.txt", "r") as file:
    logs = file.readlines()

print(logs)
```

# Step 3: Count Log Types

```python
error_count = 0
warning_count = 0
info_count = 0

for line in logs:
    if "ERROR" in line:
        error_count += 1
    elif "WARNING" in line:
        warning_count += 1
    elif "INFO" in line:
        info_count += 1

print("Errors:", error_count)
print("Warnings:", warning_count)
print("Info:", info_count)
```

# Step 4: Extract Error Messages

```python
errors = [line for line in logs if "ERROR" in line]

print(errors)
```

# Step 5: Save Report

```python
with open("report.txt", "w") as file:
    file.write(f"Errors: {error_count}\n")
    file.write(f"Warnings: {warning_count}\n")
    file.write(f"Info: {info_count}\n")
```
