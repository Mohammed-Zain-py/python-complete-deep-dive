# Why Data Cleaning?

Real-world data is:

* Messy
* Inconsistent
* Incomplete

Data cleaning is the **first step in data analysis**.

```python
import csv
```

# Step 1: Sample Dirty Data

```python
data = [
    ["name", "age", "city"],
    ["Zain", "21", "Bangalore"],
    ["Ali", "", "Mumbai"],        # missing age
    ["Zain", "21", "Bangalore"],  # duplicate
    ["John", "25", "delhi"]       # lowercase city
]
```

# Step 2: Remove Duplicates

```python
unique_data = []

for row in data:
    if row not in unique_data:
        unique_data.append(row)

print(unique_data)
```

# Step 3: Handle Missing Values

```python
clean_data = []

for row in unique_data:
    if "" in row:
        continue   # skip rows with missing values
    clean_data.append(row)

print(clean_data)
```

# Step 4: Standardize Data

```python
for i in range(1, len(clean_data)):
    clean_data[i][2] = clean_data[i][2].capitalize()

print(clean_data)
```

# Step 5: Save Clean Data to CSV

```python
with open("clean_data.csv", "w", newline="") as file:
    writer = csv.writer(file)
    writer.writerows(clean_data)
```
